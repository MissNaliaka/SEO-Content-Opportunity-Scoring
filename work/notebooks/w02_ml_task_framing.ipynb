The research question for lane 2 is: Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?

The question is more of a ranking question. The goal is to create an ordered list of the pages with a priority score so that a reviewer knows which pages to prioritize for refreshing, monitoring, expansion, protection, or pruning.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

What I would predict: I would predict a per-page score that represent how urgently a page should be reviewed for refresh, expansion, protection, pruning, or monitoring. This is based on lane 2.

Currently, the label is based on the starter dataset which includes the 'is_declining_label' derived from trend_direction == "down". This is a defined rule, but not an observed future outcome. Trend direction was computed from trend_pct at the same point in time as the rest of the row's features. Therefore, the model would be learning from a label that already exists in the present, not one confirmed by what actually happened later.

What the ideal version would look like: A genuinely observed outcome would track the same page forward — for example, whether a page's impressions or position are still declining 30-60 days after the snapshot. That would require repeated measurements over time, which this single-snapshot dataset doesn't provide. For the next several weeks, I will work with the proxy label while being explicit that it's a stand-in, not a confirmed future result.

A limitation the data itself reveals: Even the proxy has a weakness. 43.8% of pages are labeled "declining" while still receiving meaningful impressions (≥100 in the last 90 days) — a binary label collapses these into the same bucket as pages that are declining and nearly invisible, even though a reviewer would treat them very differently. This is part of why the task is better framed as a ranking/score, not a strict classification: the binary label alone loses information that matters for prioritization.
